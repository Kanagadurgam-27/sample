```python
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------------
# 1. Read the dataset
# ---------------------------------------------------

data = pd.read_csv("data_core.csv")

print("Dataset:")
print(data.head())

print("\nDataset Information:")
print(data.info())


# ---------------------------------------------------
# 2. BOX PLOT
# ---------------------------------------------------

plt.figure(figsize=(10, 6))

data.boxplot(
    column=[
        "Temparature",
        "Humidity",
        "Moisture",
        "Nitrogen",
        "Potassium",
        "Phosphorous"
    ]
)

plt.title("Box Plot of Numerical Variables")
plt.xlabel("Parameters")
plt.ylabel("Values")
plt.xticks(rotation=20)
plt.grid(True)

plt.show()


# ---------------------------------------------------
# 3. DATA PLOT / LINE PLOT
# ---------------------------------------------------

# Select first 100 records
sample = data.head(100)

plt.figure(figsize=(12, 6))

plt.plot(
    sample.index,
    sample["Nitrogen"],
    label="Nitrogen"
)

plt.plot(
    sample.index,
    sample["Potassium"],
    label="Potassium"
)

plt.plot(
    sample.index,
    sample["Phosphorous"],
    label="Phosphorous"
)

plt.title("Nutrient Data Plot")
plt.xlabel("Record Number")
plt.ylabel("Nutrient Value")

plt.legend()
plt.grid(True)

plt.show()


# ---------------------------------------------------
# 4. PIE CHART
# ---------------------------------------------------

fertilizer_count = data["Fertilizer Name"].value_counts()

# Select top 10 fertilizer categories
top10 = fertilizer_count.head(10)

plt.figure(figsize=(9, 9))

plt.pie(
    top10.values,
    labels=top10.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Fertilizer Name Distribution")

plt.show()


# ---------------------------------------------------
# 5. SCATTER PLOT
# ---------------------------------------------------

plt.figure(figsize=(9, 6))

plt.scatter(
    data["Nitrogen"],
    data["Phosphorous"],
    alpha=0.5
)

plt.title("Scatter Plot: Nitrogen vs Phosphorous")
plt.xlabel("Nitrogen")
plt.ylabel("Phosphorous")

plt.grid(True)

plt.show()
```
